```python
import torch
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(42)

text = "the cat is sitting on the mat the dog is playing in the garden the cat is playing"
words = text.split()
vocab = sorted(set(words))

word_to_id = {w: i for i, w in enumerate(vocab)}
id_to_word = {i: w for w, i in word_to_id.items()}

data = torch.tensor([word_to_id[w] for w in words])

class TransformerModel(nn.Module):
    def __init__(self, vocab_size, d_model=32):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, d_model)
        self.pos = nn.Parameter(torch.randn(1, 50, d_model))
        self.attention = nn.MultiheadAttention(d_model, 2, batch_first=True)
        self.fc = nn.Linear(d_model, vocab_size)

    def forward(self, x):
        n = x.size(1)
        x = self.embed(x) + self.pos[:, :n, :]
        mask = torch.triu(
            torch.ones(n, n, device=x.device), diagonal=1
        ).bool()
        x, _ = self.attention(x, x, x, attn_mask=mask)
        return self.fc(x)

model = TransformerModel(len(vocab))
optimizer = optim.Adam(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss()

x = data[:-1].unsqueeze(0)
y = data[1:].unsqueeze(0)

for epoch in range(300):
    model.train()
    output = model(x)
    loss = loss_fn(output.reshape(-1, len(vocab)), y.reshape(-1))
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

def predict_next_word(sentence):
    model.eval()
    tokens = sentence.lower().split()

    if not tokens or any(w not in word_to_id for w in tokens):
        return "Please enter words from the training vocabulary: " + ", ".join(vocab)

    ids = torch.tensor([word_to_id[w] for w in tokens]).unsqueeze(0)

    with torch.no_grad():
        output = model(ids)

    next_id = output[0, -1].argmax().item()
    return id_to_word[next_id]

print("\nNext-Token Prediction Model")
sentence = input("Enter a sentence: ")
print("Input:", sentence)
print("Predicted next word:", predict_next_word(sentence))
```
